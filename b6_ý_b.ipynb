```python
import numpy as np
import matplotlib.pyplot as plt

def f(x): return x*2 - 4
def df(x): return 2*x

x_n = 5  # Điểm bắt đầu
history = [x_n]

for i in range(5):
    x_n = x_n - f(x_n)/df(x_n)
    history.append(x_n)

# Vẽ biểu đồ
x_range = np.linspace(-1, 6, 100)
plt.plot(x_range, f(x_range), label='f(x) = x^2 - 4')
plt.axhline(0, color='black', lw=1)
for i, x in enumerate(history):
    plt.plot(x, f(x), 'ro')
    plt.annotate(f'x{i}', (x, f(x)))
plt.legend()
plt.title("Minh họa phương pháp Newton")
plt.show()
```
