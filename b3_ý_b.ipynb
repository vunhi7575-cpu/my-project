```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Khởi tạo dữ liệu từ hình ảnh
data = {
    'Doanh_thu': [3, 6, 9, 10, 12, 15, 18, 20],
    'MKT': [0.5, 1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0],
    'Loi_nhuan': [1.1, 2.0, 3.2, 3.4, 3.9, 5.0, 6.2, 6.6]
}
df = pd.DataFrame(data)

# 2. Huấn luyện mô hình hồi quy 2 biến
X = df[['Doanh_thu', 'MKT']]
y = df['Loi_nhuan']
model = LinearRegression().fit(X, y)

# 3. Dự báo cho tháng 9 (Doanh thu 25, MKT 5.0)
X_test = np.array([[25, 5.0]])
y_pred = model.predict(X_test)
print(f"Dự báo lợi nhuận tháng 9: {y_pred[0]:.2f}")

# 4. Vẽ biểu đồ minh họa (Doanh thu vs Lợi nhuận)
plt.scatter(df['Doanh_thu'], y, color='blue', label='Dữ liệu thực tế')
plt.plot(df['Doanh_thu'], model.predict(X), color='red', label='Đường hồi quy')
plt.xlabel('Doanh thu')
plt.ylabel('Lợi nhuận')
plt.legend()
plt.title('Hồi quy: Doanh thu & MKT tác động đến Lợi nhuận')
plt.show()
```
