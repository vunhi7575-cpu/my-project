```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import pandas as pd

# Assuming 'df' is already loaded and contains 'Thang' and 'Lợi nhuận'
# For demonstration, let's create a sample df if it's not defined.
# In a real scenario, you would load your actual data into 'df'.
if 'df' not in locals():
    data = {'Thang': range(1, 13), 'Loi_nhuan_y': [100, 120, 150, 130, 180, 200, 220, 210, 240, 260, 280, 300]} # Changed 'Lợi nhuận' to 'Loi_nhuan_y'
    df = pd.DataFrame(data)

# Define X and y
X = df[['Thang']]
y = df['Loi_nhuan_y'] # Changed 'Lợi nhuận' to 'Loi_nhuan_y'

# 1. Chuẩn hóa dữ liệu
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Chạy mô hình trên dữ liệu đã chuẩn hóa
model_scaled = LinearRegression()
model_scaled.fit(X_scaled, y)
y_pred_scaled = model_scaled.predict(X_scaled)

# 3. So sánh kết quả (Trực quan hóa)
plt.figure(figsize=(10, 6))
plt.plot(df['Thang'], y, 'ro-', label='Thực tế')
plt.plot(df['Thang'], y_pred_scaled, 'b--', label='Dự báo (Sau chuẩn hóa)')

plt.title("Kết quả mô hình sau khi chuẩn hóa dữ liệu (Mục 5)")
plt.xlabel("Tháng")
plt.ylabel("Lợi nhuận (VND)") # 'Lợi nhuận' is acceptable here as a display label
plt.legend()
plt.grid(True)
plt.show()

# In ra sai số để so sánh
mse_final = mean_squared_error(y, y_pred_scaled)
print(f"MSE sau khi chuẩn hóa: {mse_final:.2f}")
```
