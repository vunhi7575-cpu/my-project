```python
import numpy as np

# 1. Dữ liệu giả lập
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

# 2. Khởi tạo tham số
a, b = 0.0, 0.0
lr = 0.01  # Learning rate
epochs = 1000
n = float(len(X))

# 3. Vòng lặp Gradient Descent
for i in range(epochs):
    y_pred = a * X + b

    # Tính đạo hàm (Gradient) của hàm MSE
    da = (-2/n) * sum(X * (y - y_pred))
    db = (-2/n) * sum(y - y_pred)

    # Cập nhật a và b
    a = a - lr * da
    b = b - lr * db

print(f"Kết quả: y = {a:.2f}x + {b:.2f}")
```
