```python
# =========================
# 1. Dataset
# =========================

X = [1, 2, 3, 4, 5]
Y = [2, 4, 6, 8, 10]

n = len(X)


# =========================
# 2. Khởi tạo tham số
# =========================

w = 0
b = 0

learning_rate = 0.01
epochs = 1000


# =========================
# 3. Huấn luyện mô hình
# =========================

for epoch in range(epochs):

    # Step 1: Dự đoán
    y_pred = []
    for x in X:
        y_pred.append(w*x + b)

    # Step 2: Tính Loss (Mean Squared Error)
    loss = 0
    for i in range(n):
        loss += (Y[i] - y_pred[i])**2
    loss = loss / n

    # Step 3: Tính Gradient
    dw = 0
    db = 0

    for i in range(n):
        error = Y[i] - y_pred[i]
        dw += -2 * X[i] * error
        db += -2 * error

    dw = dw / n
    db = db / n

    # Step 4: Cập nhật tham số
    w = w - learning_rate * dw
    b = b - learning_rate * db

    # In loss mỗi 100 vòng
    if epoch % 100 == 0:
        print("Epoch:", epoch, "Loss:", loss)


# =========================
# 4. Kết quả sau khi train
# =========================

print("\nModel sau khi huấn luyện:")
print("w =", w)
print("b =", b)


# =========================
# 5. Hàm dự đoán
# =========================

def predict(x):
    return w*x + b


# =========================
# 6. Dự đoán dữ liệu mới
# =========================

x_test = 6
y_predict = predict(x_test)

print("\nDự đoán với x =", x_test)
print("y =", y_predict)
```
