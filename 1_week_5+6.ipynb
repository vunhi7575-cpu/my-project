```python
# Dataset
X = [1, 2, 3, 4, 5]
Y = [2, 4, 6, 8, 10]

# Khởi tạo tham số
w = 0
b = 0

learning_rate = 0.01
epochs = 1000

n = len(X)

for epoch in range(epochs):

    # Tính dự đoán
    y_pred = []
    for x in X:
        y_pred.append(w*x + b)

    # Tính gradient
    dw = 0
    db = 0

    for i in range(n):
        error = Y[i] - y_pred[i]
        dw += -2 * X[i] * error
        db += -2 * error

    dw = dw / n
    db = db / n

    # Cập nhật tham số
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("w =", w)
print("b =", b)
```
